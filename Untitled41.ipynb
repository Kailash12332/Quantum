```python

# ============================================================
# ALABAMA FOREST DROUGHT STRESS MONITORING (2016–2024)
# FINAL RANDOM FOREST REGRESSION ANALYSIS
#
# Target: NDMI_Anomaly
#
# Model 1: Climate-only
# Model 2: Previous-month NDMI only (Persistence)
# Model 3: Climate + Previous-month NDMI (Combined)
#
# Training:   2016–2021
# Validation: 2022–2023
# Testing:    2024 (independent)
#
# Outputs:
# - Data quality report
# - Validation metrics
# - Independent test metrics
# - Observed vs predicted plots
# - Residual distribution
# - Feature importance
# - Monthly and seasonal performance
# - Predicted vs observed CSV
# - ZIP file containing all results
# ============================================================

import os
import warnings
import zipfile
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files
from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import ParameterGrid
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
from sklearn.inspection import permutation_importance

warnings.filterwarnings("ignore", category=FutureWarning)

# ------------------------------------------------------------
# 1. UPLOAD DATA
# ------------------------------------------------------------

print("Upload your final 2016–2024 CSV dataset.")
uploaded = files.upload()

csv_files = [
    name for name in uploaded
    if name.lower().endswith(".csv")
]

if len(csv_files) != 1:
    raise ValueError("Please upload exactly one CSV file.")

filename = csv_files[0]
df = pd.read_csv(filename)

output_dir = Path("Alabama_RF_Results_2016_2024")
output_dir.mkdir(exist_ok=True)

print("\nDataset loaded:", filename)
print("Original shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

# ------------------------------------------------------------
# 2. VERIFY REQUIRED COLUMNS
# ------------------------------------------------------------

climate_features = [
    "SPI_3",
    "SPI_6",
    "SPI_12",
    "SPEI_3",
    "SPEI_6",
    "SPEI_12",
    "Precipitation",
    "Temperature",
    "Soil_Moisture",
    "VPD",
    "Elevation",
    "Slope",
    "Month_sin",
    "Month_cos"
]

target = "NDMI_Anomaly"

required = [
    "Point_ID",
    "Year",
    "Month",
    target
] + climate_features

missing_columns = [
    c for c in required if c not in df.columns
]

if missing_columns:
    raise ValueError(
        "Missing required columns: "
        + str(missing_columns)
    )

# ------------------------------------------------------------
# 3. CLEAN DATA
# ------------------------------------------------------------

df = df.copy()

df["Year"] = pd.to_numeric(
    df["Year"], errors="coerce"
)

df["Month"] = pd.to_numeric(
    df["Month"], errors="coerce"
)

df = df[
    df["Year"].between(2016, 2024)
    & df["Month"].between(1, 12)
].copy()

df["Year"] = df["Year"].astype(int)
df["Month"] = df["Month"].astype(int)

df["Date"] = pd.to_datetime(
    dict(
        year=df["Year"],
        month=df["Month"],
        day=1
    )
)

# Check duplicate location-month observations
keys = ["Point_ID", "Date"]

duplicate_count = df.duplicated(
    keys, keep=False
).sum()

print("\nDuplicate point-month rows:", duplicate_count)

if duplicate_count > 0:
    df.loc[
        df.duplicated(keys, keep=False)
    ].to_csv(
        output_dir / "Duplicate_Records.csv",
        index=False
    )
    raise ValueError(
        "Duplicate observations found. "
        "Inspect Duplicate_Records.csv before modeling."
    )

numeric_columns = climate_features + [target]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col], errors="coerce"
    )
    df[col] = df[col].replace(
        [np.inf, -np.inf], np.nan
    )

df = df.dropna(subset=[target])

df = df.sort_values(
    ["Point_ID", "Date"]
).reset_index(drop=True)

# ------------------------------------------------------------
# 4. PREVIOUS-MONTH NDMI ANOMALY
# ------------------------------------------------------------

df["Previous_NDMI"] = (
    df.groupby("Point_ID")[target].shift(1)
)

previous_date = (
    df.groupby("Point_ID")["Date"].shift(1)
)

month_difference = (
    (df["Year"] - previous_date.dt.year) * 12
    + (df["Month"] - previous_date.dt.month)
)

# Do not connect nonconsecutive months
df.loc[
    month_difference != 1,
    "Previous_NDMI"
] = np.nan

# Use the same observations for all models
df_model = df.dropna(
    subset=["Previous_NDMI"]
).copy()

# ------------------------------------------------------------
# 5. DATA QUALITY REPORT
# ------------------------------------------------------------

quality = pd.DataFrame({
    "Variable": df_model.columns,
    "Missing_Count": df_model.isna().sum().values,
    "Missing_Percent": (
        df_model.isna().mean().values * 100
    )
})

quality.to_csv(
    output_dir / "Data_Quality_Report.csv",
    index=False
)

print("\nFINAL DATA SUMMARY")
print("Total observations:", len(df_model))
print("Unique forest points:", df_model["Point_ID"].nunique())
print("Years:", df_model["Year"].min(),
      "-", df_model["Year"].max())

print("\nTarget statistics:")
print(df_model[target].describe())

# Flag unusually large target values
if (
    df_model[target].min() < -2
    or df_model[target].max() > 2
):
    print(
        "\nWARNING: NDMI_Anomaly exceeds the possible "
        "range of a simple raw NDMI difference. "
        "Verify whether the target is standardized."
    )

# ------------------------------------------------------------
# 6. DEFINE MODELS
# ------------------------------------------------------------

feature_sets = {
    "Climate_Only": climate_features,
    "Persistence_Only": ["Previous_NDMI"],
    "Combined": climate_features + ["Previous_NDMI"]
}

# ------------------------------------------------------------
# 7. CHRONOLOGICAL SPLIT
# ------------------------------------------------------------

train = df_model[
    df_model["Year"].between(2016, 2021)
].copy()

valid = df_model[
    df_model["Year"].between(2022, 2023)
].copy()

test = df_model[
    df_model["Year"] == 2024
].copy()

if min(len(train), len(valid), len(test)) == 0:
    raise ValueError(
        "Training, validation, or testing set is empty."
    )

print("\nDATA SPLIT")
print("Training (2016–2021):", len(train))
print("Validation (2022–2023):", len(valid))
print("Testing (2024):", len(test))

# ------------------------------------------------------------
# 8. EVALUATION FUNCTION
# ------------------------------------------------------------

def calculate_metrics(y_true, y_pred):

    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    rmse = np.sqrt(
        mean_squared_error(y_true, y_pred)
    )

    return {
        "R2": r2_score(y_true, y_pred),
        "RMSE": rmse,
        "MAE": mean_absolute_error(y_true, y_pred)
    }

def build_model(params):

    return Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "random_forest",
            RandomForestRegressor(
                **params,
                random_state=42,
                n_jobs=-1
            )
        )
    ])

# ------------------------------------------------------------
# 9. HYPERPARAMETER TUNING
# ------------------------------------------------------------

# Moderate search grid for Google Colab
param_grid = {
    "n_estimators": [150, 300],
    "max_depth": [12, None],
    "min_samples_leaf": [2, 5],
    "max_features": [0.7]
}

best_params = {}
validation_rows = []

for name, features in feature_sets.items():

    print("\nTraining:", name)

    X_train = train[features]
    y_train = train[target]

    X_valid = valid[features]
    y_valid = valid[target]

    best_rmse = np.inf
    best_model = None
    best_configuration = None

    for params in ParameterGrid(param_grid):

        model = build_model(params)

        model.fit(X_train, y_train)

        prediction = model.predict(X_valid)

        scores = calculate_metrics(
            y_valid, prediction
        )

        if scores["RMSE"] < best_rmse:
            best_rmse = scores["RMSE"]
            best_model = model
            best_configuration = params

    best_params[name] = best_configuration

    final_validation_prediction = (
        best_model.predict(X_valid)
    )

    scores = calculate_metrics(
        y_valid,
        final_validation_prediction
    )

    validation_rows.append({
        "Model": name,
        **scores,
        "Best_Parameters": str(best_configuration)
    })

    print("Best parameters:", best_configuration)
    print("Validation R2:", round(scores["R2"], 4))
    print("Validation RMSE:", round(scores["RMSE"], 4))
    print("Validation MAE:", round(scores["MAE"], 4))

validation_results = pd.DataFrame(validation_rows)

validation_results.to_csv(
    output_dir / "Validation_Results.csv",
    index=False
)

print("\nVALIDATION RESULTS")
display(validation_results)

# ------------------------------------------------------------
# 10. TRAIN FINAL MODELS USING 2016–2023
# ------------------------------------------------------------

train_final = df_model[
    df_model["Year"].between(2016, 2023)
].copy()

test_final = df_model[
    df_model["Year"] == 2024
].copy()

final_models = {}
test_rows = []

prediction_columns = [
    "Point_ID",
    "Year",
    "Month",
    "Date",
    target
]

for col in ["latitude", "longitude"]:
    if col in test_final.columns:
        prediction_columns.append(col)

predictions = test_final[
    prediction_columns
].copy()

for name, features in feature_sets.items():

    print("\nFinal model:", name)

    model = build_model(best_params[name])

    model.fit(
        train_final[features],
        train_final[target]
    )

    predicted = model.predict(
        test_final[features]
    )

    final_models[name] = model

    predictions[name + "_Predicted"] = predicted

    scores = calculate_metrics(
        test_final[target],
        predicted
    )

    test_rows.append({
        "Model": name,
        **scores
    })

test_results = pd.DataFrame(test_rows)

test_results.to_csv(
    output_dir / "Independent_Test_2024.csv",
    index=False
)

predictions.to_csv(
    output_dir / "Predictions_2024.csv",
    index=False
)

print("\n====================================")
print("INDEPENDENT TEST RESULTS — 2024")
print("====================================")
display(test_results.round(4))

# ------------------------------------------------------------
# 11. MODEL COMPARISON FIGURE
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(9, 5))

ax.bar(
    test_results["Model"],
    test_results["R2"]
)

ax.set_ylabel("R²")
ax.set_title(
    "Random Forest Model Comparison — 2024"
)
ax.axhline(0, color="black", linewidth=0.8)

plt.xticks(rotation=15)
plt.tight_layout()

plt.savefig(
    output_dir / "Figure_1_Model_Comparison.png",
    dpi=300
)
plt.show()

# ------------------------------------------------------------
# 12. OBSERVED VS PREDICTED
# ------------------------------------------------------------

fig, axes = plt.subplots(
    1, 3, figsize=(18, 5)
)

for ax, name in zip(
    axes, feature_sets.keys()
):

    observed = test_final[target].to_numpy()

    predicted = predictions[
        name + "_Predicted"
    ].to_numpy()

    ax.scatter(
        observed,
        predicted,
        alpha=0.25,
        s=10
    )

    lower = min(
        observed.min(),
        predicted.min()
    )

    upper = max(
        observed.max(),
        predicted.max()
    )

    ax.plot(
        [lower, upper],
        [lower, upper],
        "k--",
        linewidth=1.5
    )

    score = calculate_metrics(
        observed, predicted
    )

    ax.set_title(
        f"{name}\n"
        f"R² = {score['R2']:.3f}, "
        f"RMSE = {score['RMSE']:.3f}"
    )

    ax.set_xlabel("Observed NDMI anomaly")
    ax.set_ylabel("Predicted NDMI anomaly")

plt.tight_layout()

plt.savefig(
    output_dir / "Figure_2_Observed_vs_Predicted.png",
    dpi=300
)
plt.show()

# ------------------------------------------------------------
# 13. RESIDUAL ANALYSIS
# ------------------------------------------------------------

best_model_name = (
    test_results.sort_values(
        "RMSE"
    ).iloc[0]["Model"]
)

print("\nBest test-performing model:", best_model_name)
print(
    "Note: Model selection was based on validation; "
    "test performance is reported for comparison."
)

observed = test_final[target].to_numpy()

predicted = predictions[
    best_model_name + "_Predicted"
].to_numpy()

residuals = observed - predicted

plt.figure(figsize=(9, 5))

plt.hist(
    residuals,
    bins=50,
    edgecolor="black",
    alpha=0.8
)

plt.axvline(
    0, color="black", linestyle="--"
)

plt.xlabel("Residual (Observed - Predicted)")
plt.ylabel("Frequency")
plt.title(
    f"Residual Distribution — {best_model_name}"
)

plt.tight_layout()

plt.savefig(
    output_dir / "Figure_3_Residual_Distribution.png",
    dpi=300
)
plt.show()

# ------------------------------------------------------------
# 14. PERMUTATION FEATURE IMPORTANCE
# ------------------------------------------------------------

# Calculate for combined model
combined_features = feature_sets["Combined"]

combined_model = final_models["Combined"]

importance = permutation_importance(
    combined_model,
    test_final[combined_features],
    test_final[target],
    scoring="neg_mean_squared_error",
    n_repeats=5,
    random_state=42,
    n_jobs=-1
)

importance_df = pd.DataFrame({
    "Feature": combined_features,
    "Importance": importance.importances_mean,
    "Std": importance.importances_std
})

importance_df = importance_df.sort_values(
    "Importance", ascending=False
)

importance_df.to_csv(
    output_dir / "Feature_Importance.csv",
    index=False
)

print("\nFEATURE IMPORTANCE")
display(importance_df)

plt.figure(figsize=(10, 7))

plt.barh(
    importance_df["Feature"][::-1],
    importance_df["Importance"][::-1]
)

plt.xlabel(
    "Increase in MSE after permutation"
)
plt.ylabel("Predictor")
plt.title(
    "Combined Random Forest — Feature Importance"
)

plt.tight_layout()

plt.savefig(
    output_dir / "Figure_4_Feature_Importance.png",
    dpi=300
)
plt.show()

# ------------------------------------------------------------
# 15. MONTHLY MODEL PERFORMANCE
# ------------------------------------------------------------

monthly_results = []

for name in feature_sets.keys():

    pred_col = name + "_Predicted"

    for month in sorted(predictions["Month"].unique()):

        subset = predictions[
            predictions["Month"] == month
        ]

        if len(subset) < 2:
            continue

        scores = calculate_metrics(
            subset[target],
            subset[pred_col]
        )

        monthly_results.append({
            "Model": name,
            "Month": month,
            "N": len(subset),
            **scores
        })

monthly_df = pd.DataFrame(monthly_results)

monthly_df.to_csv(
    output_dir / "Monthly_Performance_2024.csv",
    index=False
)

print("\nMONTHLY PERFORMANCE")
display(monthly_df.round(4))

plt.figure(figsize=(10, 5))

for name in feature_sets.keys():

    subset = monthly_df[
        monthly_df["Model"] == name
    ]

    plt.plot(
        subset["Month"],
        subset["RMSE"],
        marker="o",
        label=name
    )

plt.xticks(range(1, 13))
plt.xlabel("Month")
plt.ylabel("RMSE")
plt.title("Monthly Model Performance — 2024")
plt.legend()

plt.tight_layout()

plt.savefig(
    output_dir / "Figure_5_Monthly_RMSE.png",
    dpi=300
)
plt.show()

# ------------------------------------------------------------
# 16. SEASONAL MODEL PERFORMANCE
# ------------------------------------------------------------

season_map = {
    12: "Winter",
    1: "Winter",
    2: "Winter",
    3: "Spring",
    4: "Spring",
    5: "Spring",
    6: "Summer",
    7: "Summer",
    8: "Summer",
    9: "Fall",
    10: "Fall",
    11: "Fall"
}

predictions["Season"] = (
    predictions["Month"].map(season_map)
)

seasonal_rows = []

for name in feature_sets.keys():

    for season in [
        "Winter", "Spring", "Summer", "Fall"
    ]:

        subset = predictions[
            predictions["Season"] == season
        ]

        if len(subset) < 2:
            continue

        scores = calculate_metrics(
            subset[target],
            subset[name + "_Predicted"]
        )

        seasonal_rows.append({
            "Model": name,
            "Season": season,
            "N": len(subset),
            **scores
        })

seasonal_df = pd.DataFrame(seasonal_rows)

seasonal_df.to_csv(
    output_dir / "Seasonal_Performance_2024.csv",
    index=False
)

print("\nSEASONAL PERFORMANCE")
display(seasonal_df.round(4))

# ------------------------------------------------------------
# 17. PERSISTENCE VS CLIMATE IMPROVEMENT
# ------------------------------------------------------------

persistence_rmse = test_results.loc[
    test_results["Model"] == "Persistence_Only",
    "RMSE"
].iloc[0]

combined_rmse = test_results.loc[
    test_results["Model"] == "Combined",
    "RMSE"
].iloc[0]

improvement = (
    (persistence_rmse - combined_rmse)
    / persistence_rmse
) * 100

print("\n====================================")
print("CLIMATE INFORMATION ADDED VALUE")
print("====================================")

print(
    f"Persistence RMSE: {persistence_rmse:.4f}"
)

print(
    f"Combined RMSE: {combined_rmse:.4f}"
)

print(
    f"RMSE improvement: {improvement:.2f}%"
)

if improvement > 0:
    print(
        "Combined model performs better than "
        "persistence on the 2024 test set."
    )
else:
    print(
        "Climate predictors did not improve "
        "over persistence on the 2024 test set."
    )

# ------------------------------------------------------------
# 18. SAVE CLEANED DATA
# ------------------------------------------------------------

df_model.to_csv(
    output_dir / "Cleaned_Model_Data_2016_2024.csv",
    index=False
)

# ------------------------------------------------------------
# 19. SAVE METHODOLOGY NOTES
# ------------------------------------------------------------

notes = """
Alabama Forest Drought Random Forest Analysis
Period: 2016–2024

Target:
NDMI_Anomaly

Training:
2016–2021

Validation:
2022–2023

Independent testing:
2024

Models:
1. Climate-only
2. Previous-month NDMI anomaly
3. Combined climate + previous-month NDMI anomaly

Random Forest parameters were selected using
validation RMSE.

Final models were trained on 2016–2023
and tested on 2024.

The same point-month observations were used
for all three model comparisons.

Previous-month NDMI was calculated only when
the preceding observation was exactly one
calendar month earlier.

Important limitations:
1. Verify whether NDMI_Anomaly is standardized.
2. Verify how the NDMI reference climatology
   was calculated.
3. Same-month climate predictors make this
   a contemporaneous estimation model,
   not a true advance forecast.
4. Correlated SPI/SPEI variables can affect
   interpretation of feature importance.
5. Nearby sampling locations may not be
   spatially independent.
6. The model demonstrates predictive
   associations, not causation.
"""

with open(
    output_dir / "Methodology_Notes.txt",
    "w"
) as f:
    f.write(notes)

# ------------------------------------------------------------
# 20. ZIP ALL RESULTS
# ------------------------------------------------------------

zip_name = "Alabama_RF_Results_2016_2024.zip"

with zipfile.ZipFile(
    zip_name, "w", zipfile.ZIP_DEFLATED
) as zipf:

    for file_path in output_dir.iterdir():

        if file_path.is_file():

            zipf.write(
                file_path,
                arcname=file_path.name
            )

print("\n====================================")
print("ANALYSIS COMPLETED")
print("====================================")

print("Results folder:", output_dir)
print("ZIP file:", zip_name)

files.download(zip_name)
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = df.copy()

required = [
    "NDMI",
    "NDMI_Train_Mean",
    "NDMI_Train_SD",
    "NDMI_Anomaly",
    "Year"
]

missing = [c for c in required if c not in data.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

# Calculate raw anomaly
data["NDMI_Raw_Anomaly"] = (
    data["NDMI"] - data["NDMI_Train_Mean"]
)

# Recalculate standardized anomaly
data["NDMI_Z_Recalculated"] = np.where(
    data["NDMI_Train_SD"] > 0,
    data["NDMI_Raw_Anomaly"] / data["NDMI_Train_SD"],
    np.nan
)

# Compare calculated and existing standardized anomalies
difference = (
    data["NDMI_Z_Recalculated"] -
    data["NDMI_Anomaly"]
).abs()

print("Maximum difference:", difference.max())
print("Median difference:", difference.median())

# Compare yearly distributions
summary = data.groupby("Year")[
    ["NDMI_Anomaly", "NDMI_Raw_Anomaly"]
].agg(["mean", "std", "min", "max"])

display(summary.round(4))

# Examine 2024 distributions
subset = data[data["Year"] == 2024]

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].hist(
    subset["NDMI_Anomaly"].dropna(),
    bins=60,
    edgecolor="black"
)
axes[0].set_title("2024 Standardized NDMI Anomaly")
axes[0].set_xlabel("Standardized anomaly")

axes[1].hist(
    subset["NDMI_Raw_Anomaly"].dropna(),
    bins=60,
    edgecolor="black"
)
axes[1].set_title("2024 Raw NDMI Anomaly")
axes[1].set_xlabel("Raw NDMI anomaly")

plt.tight_layout()
plt.show()

# Examine baseline SD
print("\nReference standard deviation quantiles:")
print(data["NDMI_Train_SD"].quantile(
    [0, 0.01, 0.05, 0.10, 0.25, 0.50, 0.75, 1]
))

# Save results without changing the original dataset
data.to_csv(
    "Alabama_NDMI_Anomaly_Diagnostic_2016_2024.csv",
    index=False
)
```

```python

# ==========================================================
# ALABAMA FOREST DROUGHT STUDY (2016–2024)
# RANDOM FOREST USING RAW NDMI ANOMALIES
# ==========================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import zipfile
from pathlib import Path

from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.model_selection import ParameterGrid
from sklearn.metrics import (
    r2_score, mean_squared_error, mean_absolute_error
)
from sklearn.inspection import permutation_importance
from google.colab import files

# ----------------------------------------------------------
# 1. LOAD PREPARED DATA
# ----------------------------------------------------------

if "raw_df" in globals():
    data = raw_df.copy()
else:
    print("Upload Alabama_Raw_NDMI_Model_2016_2024.csv")
    uploaded = files.upload()
    filename = next(
        n for n in uploaded if n.lower().endswith(".csv")
    )
    data = pd.read_csv(filename)

data["Year"] = pd.to_numeric(data["Year"])
data["Month"] = pd.to_numeric(data["Month"])

data = data[
    data["Year"].between(2016, 2024)
].copy()

data["Date"] = pd.to_datetime(
    dict(year=data["Year"], month=data["Month"], day=1)
)

data = data.sort_values(
    ["Point_ID", "Date"]
).reset_index(drop=True)

# Recalculate target from observed NDMI and reference mean
data["NDMI_Raw_Anomaly"] = (
    data["NDMI"] - data["NDMI_Train_Mean"]
)

# Recalculate previous-month raw anomaly
data["Previous_Raw_NDMI"] = (
    data.groupby("Point_ID")["NDMI_Raw_Anomaly"].shift(1)
)

previous_date = data.groupby("Point_ID")["Date"].shift(1)

month_gap = (
    (data["Year"] - previous_date.dt.year) * 12
    + data["Month"] - previous_date.dt.month
)

data.loc[
    month_gap != 1, "Previous_Raw_NDMI"
] = np.nan

# ----------------------------------------------------------
# 2. DEFINE VARIABLES
# ----------------------------------------------------------

target = "NDMI_Raw_Anomaly"

climate_features = [
    "SPI_3", "SPI_6", "SPI_12",
    "SPEI_3", "SPEI_6", "SPEI_12",
    "Precipitation", "Temperature",
    "Soil_Moisture", "VPD",
    "Elevation", "Slope",
    "Month_sin", "Month_cos"
]

feature_sets = {
    "Climate_Only": climate_features,
    "Persistence_Only": ["Previous_Raw_NDMI"],
    "Combined": climate_features + ["Previous_Raw_NDMI"]
}

required = list(set(
    climate_features + [
        target, "Previous_Raw_NDMI",
        "Point_ID", "Year", "Month"
    ]
))

missing = [c for c in required if c not in data.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

if data.duplicated(["Point_ID", "Date"]).any():
    raise ValueError("Duplicate point-month observations detected.")

for col in climate_features + [target, "Previous_Raw_NDMI"]:
    data[col] = pd.to_numeric(
        data[col], errors="coerce"
    ).replace([np.inf, -np.inf], np.nan)

# Fair comparison: same observations for all models
data = data.dropna(
    subset=[target, "Previous_Raw_NDMI"]
).copy()

# ----------------------------------------------------------
# 3. CHRONOLOGICAL SPLIT
# ----------------------------------------------------------

train = data[data["Year"].between(2016, 2021)].copy()
valid = data[data["Year"].between(2022, 2023)].copy()
test = data[data["Year"] == 2024].copy()

print("Training:", len(train))
print("Validation:", len(valid))
print("2024 evaluation:", len(test))

if min(len(train), len(valid), len(test)) == 0:
    raise ValueError("One of the data splits is empty.")

# ----------------------------------------------------------
# 4. MODEL FUNCTIONS
# ----------------------------------------------------------

def get_metrics(y, pred):
    return {
        "R2": r2_score(y, pred),
        "RMSE": np.sqrt(mean_squared_error(y, pred)),
        "MAE": mean_absolute_error(y, pred)
    }

def create_model(params):
    return Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("rf", RandomForestRegressor(
            **params,
            random_state=42,
            n_jobs=-1
        ))
    ])

param_grid = {
    "n_estimators": [150, 300],
    "max_depth": [12, None],
    "min_samples_leaf": [2, 5],
    "max_features": [0.7]
}

# ----------------------------------------------------------
# 5. VALIDATION AND HYPERPARAMETER TUNING
# ----------------------------------------------------------

best_params = {}
validation_rows = []

for name, features in feature_sets.items():

    print("\nTraining:", name)

    best_rmse = np.inf
    best_model = None
    best_config = None

    for params in ParameterGrid(param_grid):

        model = create_model(params)
        model.fit(train[features], train[target])

        pred = model.predict(valid[features])
        score = get_metrics(valid[target], pred)

        if score["RMSE"] < best_rmse:
            best_rmse = score["RMSE"]
            best_model = model
            best_config = params

    best_params[name] = best_config

    pred = best_model.predict(valid[features])
    score = get_metrics(valid[target], pred)

    validation_rows.append({
        "Model": name,
        **score,
        "Parameters": str(best_config)
    })

    print(
        f"R2={score['R2']:.4f}, "
        f"RMSE={score['RMSE']:.4f}, "
        f"MAE={score['MAE']:.4f}"
    )

validation_results = pd.DataFrame(validation_rows)

print("\nVALIDATION RESULTS")
display(validation_results.round(4))

selected_model = validation_results.loc[
    validation_results["RMSE"].idxmin(), "Model"
]

print("\nBest model selected by validation:", selected_model)

# ----------------------------------------------------------
# 6. FINAL FIT: 2016–2023; EVALUATE ON 2024
# ----------------------------------------------------------

train_final = data[data["Year"] <= 2023].copy()
test_final = data[data["Year"] == 2024].copy()

final_models = {}
test_rows = []

cols = ["Point_ID", "Year", "Month", target]
for c in ["latitude", "longitude"]:
    if c in test_final.columns:
        cols.append(c)

predictions = test_final[cols].copy()

for name, features in feature_sets.items():

    model = create_model(best_params[name])

    model.fit(
        train_final[features],
        train_final[target]
    )

    pred = model.predict(test_final[features])

    final_models[name] = model
    predictions[name + "_Predicted"] = pred

    score = get_metrics(test_final[target], pred)
    test_rows.append({"Model": name, **score})

test_results = pd.DataFrame(test_rows)

print("\n2024 SENSITIVITY EVALUATION")
display(test_results.round(4))

# ----------------------------------------------------------
# 7. OBSERVED VS PREDICTED
# ----------------------------------------------------------

fig, axes = plt.subplots(1, 3, figsize=(17, 5))

for ax, name in zip(axes, feature_sets):

    y = test_final[target].to_numpy()
    pred = predictions[name + "_Predicted"].to_numpy()

    ax.scatter(y, pred, alpha=0.25, s=12)

    lo = min(y.min(), pred.min())
    hi = max(y.max(), pred.max())

    ax.plot([lo, hi], [lo, hi], "k--")

    score = get_metrics(y, pred)

    ax.set_title(
        f"{name}\nR2={score['R2']:.3f}, "
        f"RMSE={score['RMSE']:.4f}"
    )
    ax.set_xlabel("Observed raw NDMI anomaly")
    ax.set_ylabel("Predicted raw NDMI anomaly")

plt.tight_layout()

# ----------------------------------------------------------
# 8. FEATURE IMPORTANCE
# ----------------------------------------------------------

# Use validation-selected model.
features = feature_sets[selected_model]
model = final_models[selected_model]

# Evaluate permutation importance on 2024 as
# descriptive sensitivity analysis, not tuning.
importance = permutation_importance(
    model,
    test_final[features],
    test_final[target],
    scoring="neg_mean_squared_error",
    n_repeats=5,
    random_state=42,
    n_jobs=-1
)

importance_df = pd.DataFrame({
    "Feature": features,
    "Importance": importance.importances_mean,
    "Std": importance.importances_std
}).sort_values("Importance", ascending=False)

print("\nFEATURE IMPORTANCE")
display(importance_df)

# ----------------------------------------------------------
# 9. SAVE RESULTS AND FIGURES
# ----------------------------------------------------------

output_dir = Path("Alabama_RF_Raw_NDMI_Results")
output_dir.mkdir(exist_ok=True)

fig.savefig(
    output_dir / "Observed_vs_Predicted.png",
    dpi=300, bbox_inches="tight"
)
plt.show()

plt.figure(figsize=(9, 6))
plt.barh(
    importance_df["Feature"][::-1],
    importance_df["Importance"][::-1]
)
plt.xlabel("Increase in MSE after permutation")
plt.title(f"Feature Importance: {selected_model}")
plt.tight_layout()
plt.savefig(
    output_dir / "Feature_Importance.png",
    dpi=300
)
plt.show()

validation_results.to_csv(
    output_dir / "Validation_Metrics.csv",
    index=False
)

test_results.to_csv(
    output_dir / "Test_Metrics_2024.csv",
    index=False
)

predictions.to_csv(
    output_dir / "Predictions_2024.csv",
    index=False
)

importance_df.to_csv(
    output_dir / "Feature_Importance.csv",
    index=False
)

# ----------------------------------------------------------
# 10. DOWNLOAD ALL RESULTS
# ----------------------------------------------------------

zip_name = "Alabama_RF_Raw_NDMI_Results.zip"

with zipfile.ZipFile(zip_name, "w", zipfile.ZIP_DEFLATED) as z:
    for p in output_dir.iterdir():
        if p.is_file():
            z.write(p, arcname=p.name)

print("\nAnalysis complete.")
print("Validation-selected model:", selected_model)
print("Results saved:", zip_name)

files.download(zip_name)
```

```python
# ============================================================
# ALABAMA FOREST DROUGHT MACHINE LEARNING PROJECT
# COMPLETE MODEL PERFORMANCE EVALUATION
#
# Models:
# 1. Climate-only Random Forest
# 2. Persistence-only Random Forest
# 3. Combined Random Forest
#
# Metrics:
# R2, RMSE, MSE, MAE, NMSE, NRMSE
# ============================================================

!pip -q install openpyxl

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)

from google.colab import files

# ============================================================
# STEP 1: UPLOAD YOUR MODEL PREDICTION CSV
# ============================================================

print("Upload your CSV containing observed and predicted NDMI (e.g., Predictions_2024.csv).")

uploaded = files.upload()

filename = next(iter(uploaded))
df = pd.read_csv(filename)

print("\nDataset shape:", df.shape)
print("\nAvailable columns:")
print(df.columns.tolist())

display(df.head())

# ============================================================
# STEP 2: IDENTIFY YOUR COLUMNS
# ============================================================
# Updated to match columns generated in Predictions_2024.csv

actual_col = "NDMI_Raw_Anomaly"

model_columns = {
    "Climate-only": "Climate_Only_Predicted",
    "Persistence-only": "Persistence_Only_Predicted",
    "Combined": "Combined_Predicted"
}

required = [actual_col] + list(model_columns.values())

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(
        f"Missing columns: {missing}\n"
        "Please ensure you uploaded the correct model predictions file (e.g., Predictions_2024.csv) "
        "or update the column names in STEP 2 to match your CSV."
    )

# Convert required columns to numeric
for col in required:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Retain rows with all predictions available
evaluation = df.dropna(subset=required).copy()

print("\nValid evaluation records:", len(evaluation))

if len(evaluation) < 2:
    raise ValueError("At least two valid records are required.")

# ============================================================
# STEP 3: CALCULATE MODEL PERFORMANCE METRICS
# ============================================================

def evaluate_model(y_true, y_pred):

    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)

    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)

    # Variance-normalized MSE
    observed_variance = np.mean(
        (y_true - np.mean(y_true)) ** 2
    )

    nmse = (
        mse / observed_variance
        if observed_variance > 0
        else np.nan
    )

    # Range-normalized RMSE
    observed_range = np.max(y_true) - np.min(y_true)

    nrmse = (
        rmse / observed_range
        if observed_range > 0
        else np.nan
    )

    return {
        "R2": r2,
        "RMSE": rmse,
        "MSE": mse,
        "MAE": mae,
        "NMSE": nmse,
        "NRMSE": nrmse,
        "NRMSE_percent": nrmse * 100
    }

# ============================================================
# STEP 4: EVALUATE ALL THREE MODELS
# ============================================================

y_true = evaluation[actual_col].to_numpy()

results = []

for model_name, pred_col in model_columns.items():

    y_pred = evaluation[pred_col].to_numpy()

    metrics = evaluate_model(y_true, y_pred)
    metrics["Model"] = model_name
    metrics["N"] = len(y_true)

    results.append(metrics)

results_df = pd.DataFrame(results)

results_df = results_df[
    [
        "Model", "N", "R2", "RMSE",
        "MSE", "MAE", "NMSE",
        "NRMSE", "NRMSE_percent"
    ]
]

print("\nFINAL MODEL PERFORMANCE RESULTS")
display(results_df.round(6))

# ============================================================
# STEP 5: VERIFY NMSE CALCULATION
# ============================================================

print("\nChecking NMSE = 1 - R2")

for _, row in results_df.iterrows():

    difference = abs(
        row["NMSE"] - (1 - row["R2"])
    )

    print(
        row["Model"],
        "Difference:",
        round(difference, 12)
    )

# ============================================================
# STEP 6: CREATE OBSERVED VS PREDICTED PLOTS
# ============================================================

fig, axes = plt.subplots(
    1, 3, figsize=(17, 5.5)
)

all_predictions = [
    evaluation[c].to_numpy()
    for c in model_columns.values()
]

minimum = min(
    np.min(y_true),
    *(np.min(p) for p in all_predictions)
)

maximum = max(
    np.max(y_true),
    *(np.max(p) for p in all_predictions)
)

for ax, (model_name, pred_col) in zip(
    axes, model_columns.items()
):

    y_pred = evaluation[pred_col].to_numpy()

    row = results_df[
        results_df["Model"] == model_name
    ].iloc[0]

    ax.scatter(
        y_true,
        y_pred,
        alpha=0.3,
        s=12
    )

    ax.plot(
        [minimum, maximum],
        [minimum, maximum],
        "r--",
        linewidth=1.5
    )

    ax.set_title(
        f"{model_name}\n"
        f"R² = {row['R2']:.3f}, "
        f"RMSE = {row['RMSE']:.4f}"
    )

    ax.set_xlabel("Observed NDMI anomaly")
    ax.set_ylabel("Predicted NDMI anomaly")

    ax.set_xlim(minimum, maximum)
    ax.set_ylim(minimum, maximum)
    ax.grid(alpha=0.25)

plt.tight_layout()

plt.savefig(
    "RF_Observed_vs_Predicted.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# STEP 7: MODEL COMPARISON BAR CHART
# ============================================================

metrics_to_plot = [
    "R2", "RMSE", "MSE", "NMSE"
]

for metric in metrics_to_plot:

    fig, ax = plt.subplots(figsize=(8, 5))

    bars = ax.bar(
        results_df["Model"],
        results_df[metric]
    )

    ax.set_title(
        f"Random Forest Model Comparison: {metric}"
    )

    ax.set_ylabel(metric)
    ax.grid(axis="y", alpha=0.25)

    for bar, value in zip(
        bars, results_df[metric]
    ):
        ax.annotate(
            f"{value:.4f}",
            (
                bar.get_x() + bar.get_width()/2,
                bar.get_height()
            ),
            xytext=(0, 5 if value >= 0 else -15),
            textcoords="offset points",
            ha="center"
        )

    plt.tight_layout()

    plt.savefig(
        f"RF_Comparison_{metric}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

# ============================================================
# STEP 8: CALCULATE IMPROVEMENT OVER PERSISTENCE
# ============================================================

persistence = results_df[
    results_df["Model" ] == "Persistence-only"
].iloc[0]

combined = results_df[
    results_df["Model"] == "Combined"
].iloc[0]

r2_gain = combined["R2"] - persistence["R2"]

rmse_reduction = (
    (persistence["RMSE"] - combined["RMSE"])
    / persistence["RMSE"]
) * 100

mse_reduction = (
    (persistence["MSE"] - combined["MSE"])
    / persistence["MSE"]
) * 100

print("\nIMPROVEMENT OVER PERSISTENCE MODEL")
print(f"R2 gain: {r2_gain:.4f}")
print(f"RMSE reduction: {rmse_reduction:.2f}%")
print(f"MSE reduction: {mse_reduction:.2f}%")

# ============================================================
# STEP 9: EXPORT RESULTS TO EXCEL
# ============================================================

output_excel = "Alabama_RF_Model_Performance.xlsx"

with pd.ExcelWriter(
    output_excel,
    engine="openpyxl"
) as writer:

    results_df.to_excel(
        writer,
        sheet_name="Model_Performance",
        index=False
    )

    evaluation.to_excel(
        writer,
        sheet_name="Evaluation_Data",
        index=False
    )

print("\nExcel file saved:", output_excel)

# ============================================================
# STEP 10: DOWNLOAD OUTPUT FILES
# ============================================================

files.download(output_excel)
files.download("RF_Observed_vs_Predicted.png")

for metric in metrics_to_plot:
    files.download(f"RF_Comparison_{metric}.png")

print("\nALL CALCULATIONS COMPLETED SUCCESSFULLY.")
```

```python

# ============================================================
# ALABAMA FOREST DROUGHT STUDY (2016–2024)
# SPI vs SPEI CORRELATION ANALYSIS
# Pearson + Spearman + Scatterplots + Heatmap
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr, spearmanr
from google.colab import files
from pathlib import Path
import zipfile

# ------------------------------------------------------------
# 1. LOAD DATA
# ------------------------------------------------------------

if "raw_df" in globals():
    df = raw_df.copy()
    print("Using existing raw_df")
else:
    print("Upload your Alabama master CSV file")
    uploaded = files.upload()
    filename = next(
        name for name in uploaded
        if name.lower().endswith(".csv")
    )
    df = pd.read_csv(filename)

df.columns = df.columns.str.strip()

df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df["Month"] = pd.to_numeric(df["Month"], errors="coerce")

df = df[df["Year"].between(2016, 2024)].copy()

# ------------------------------------------------------------
# 2. DEFINE SPI AND SPEI VARIABLES
# ------------------------------------------------------------

pairs = [
    ("SPI_3", "SPEI_3", "3-month"),
    ("SPI_6", "SPEI_6", "6-month"),
    ("SPI_12", "SPEI_12", "12-month")
]

columns = [
    "SPI_3", "SPI_6", "SPI_12",
    "SPEI_3", "SPEI_6", "SPEI_12"
]

missing = [c for c in columns if c not in df.columns]

if missing:
    raise ValueError(f"Missing columns: {missing}")

for col in columns:
    df[col] = pd.to_numeric(
        df[col], errors="coerce"
    ).replace([np.inf, -np.inf], np.nan)

output_dir = Path("Alabama_SPI_SPEI_Correlation")
output_dir.mkdir(exist_ok=True)

# ------------------------------------------------------------
# 3. CHECK DATA STRUCTURE
# ------------------------------------------------------------

print("Total records:", len(df))
print("Available years:", sorted(df["Year"].dropna().unique()))

# Important:
# If SPI/SPEI values repeat across forest points within
# the same month, repeated values should not be treated
# as independent observations.
#
# Calculate correlations using unique monthly climate
# records. For spatially varying values, this section
# averages values across available forest locations,
# yielding a statewide monthly summary.
#
# This is a descriptive statewide analysis, not a
# location-specific correlation analysis.

monthly = (
    df.groupby(["Year", "Month"], as_index=False)[columns]
      .mean()
      .sort_values(["Year", "Month"])
)

print("Unique year-month records:", len(monthly))

# ------------------------------------------------------------
# 4. PEARSON AND SPEARMAN CORRELATIONS
# ------------------------------------------------------------

results = []

for spi, spei, scale in pairs:

    subset = monthly[[spi, spei]].dropna()

    if len(subset) < 3:
        print(f"Insufficient data for {scale}")
        continue

    if (
        subset[spi].nunique() < 2 or
        subset[spei].nunique() < 2
    ):
        print(f"Constant data for {scale}")
        continue

    pearson_r, pearson_p = pearsonr(
        subset[spi], subset[spei]
    )

    spearman_r, spearman_p = spearmanr(
        subset[spi], subset[spei]
    )

    results.append({
        "Timescale": scale,
        "SPI": spi,
        "SPEI": spei,
        "N_months": len(subset),
        "Pearson_r": pearson_r,
        "Pearson_p": pearson_p,
        "Spearman_rho": spearman_r,
        "Spearman_p": spearman_p
    })

results_df = pd.DataFrame(results)

print("\nSPI vs SPEI CORRELATION RESULTS")
print(results_df.round(4).to_string(index=False))

results_df.to_csv(
    output_dir / "SPI_SPEI_Correlation_Results.csv",
    index=False
)

# ------------------------------------------------------------
# 5. SCATTERPLOTS
# ------------------------------------------------------------

fig, axes = plt.subplots(
    1, 3, figsize=(16, 5)
)

for ax, (spi, spei, scale) in zip(axes, pairs):

    subset = monthly[[spi, spei]].dropna()

    sns.regplot(
        data=subset,
        x=spi,
        y=spei,
        ax=ax,
        scatter_kws={
            "alpha": 0.55,
            "s": 30
        },
        line_kws={"color": "red"}
    )

    if (
        len(subset) >= 3
        and subset[spi].nunique() > 1
        and subset[spei].nunique() > 1
    ):
        r, _ = pearsonr(subset[spi], subset[spei])
        ax.set_title(f"{scale}: Pearson r = {r:.3f}")
    else:
        ax.set_title(f"{scale}: insufficient variation")

    ax.set_xlabel(spi)
    ax.set_ylabel(spei)
    ax.grid(alpha=0.2)

plt.suptitle(
    "SPI vs SPEI Correlation in Alabama (2016–2024)",
    fontsize=15
)
plt.tight_layout()

plt.savefig(
    output_dir / "SPI_SPEI_Scatterplots.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# ------------------------------------------------------------
# 6. CORRELATION HEATMAP
# ------------------------------------------------------------

corr_matrix = monthly[columns].corr(
    method="pearson"
)

plt.figure(figsize=(9, 7))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="RdBu_r",
    center=0,
    vmin=-1,
    vmax=1,
    square=True
)

plt.title(
    "Pearson Correlation Matrix: SPI and SPEI"
)

plt.tight_layout()

plt.savefig(
    output_dir / "SPI_SPEI_Heatmap.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

corr_matrix.to_csv(
    output_dir / "SPI_SPEI_Correlation_Matrix.csv"
)

# ------------------------------------------------------------
# 7. SEASONAL CORRELATION
# ------------------------------------------------------------

season_map = {
    12: "Winter", 1: "Winter", 2: "Winter",
    3: "Spring", 4: "Spring", 5: "Spring",
    6: "Summer", 7: "Summer", 8: "Summer",
    9: "Autumn", 10: "Autumn", 11: "Autumn"
}

monthly["Season"] = monthly["Month"].map(season_map)

seasonal_results = []

for season in ["Winter", "Spring", "Summer", "Autumn"]:

    season_df = monthly[
        monthly["Season"] == season
    ]

    for spi, spei, scale in pairs:

        sub = season_df[[spi, spei]].dropna()

        if (
            len(sub) < 3
            or sub[spi].nunique() < 2
            or sub[spei].nunique() < 2
        ):
            continue

        r, p = pearsonr(sub[spi], sub[spei])

        seasonal_results.append({
            "Season": season,
            "Timescale": scale,
            "N_months": len(sub),
            "Pearson_r": r,
            "p_value": p
        })

seasonal_df = pd.DataFrame(seasonal_results)

print("\nSEASONAL CORRELATIONS")
print(seasonal_df.round(4).to_string(index=False))

seasonal_df.to_csv(
    output_dir / "Seasonal_SPI_SPEI_Correlations.csv",
    index=False
)

# ------------------------------------------------------------
# 8. SAVE MONTHLY DATA
# ------------------------------------------------------------

monthly.to_csv(
    output_dir / "Monthly_SPI_SPEI_Data.csv",
    index=False
)

# ------------------------------------------------------------
# 9. ZIP AND DOWNLOAD RESULTS
# ------------------------------------------------------------

zip_name = "Alabama_SPI_SPEI_Correlation_Results.zip"

with zipfile.ZipFile(
    zip_name, "w", zipfile.ZIP_DEFLATED
) as z:

    for file in output_dir.iterdir():
        if file.is_file():
            z.write(file, arcname=file.name)

print("\nAnalysis complete!")
print("Results saved:", zip_name)

files.download(zip_name)
```
